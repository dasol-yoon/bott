Temporary notebook for restructuring the filename / folder structures.

```python
import os
import shutil
import re
from tqdm import tqdm
```

```python
# Define source and target directories
source_dir = '/Users/dash/Documents/Cornell/BOtt/simulation/'
target_dir = '/Users/dash/Documents/Cornell/BOtt/renamed_simulation/' 

# Ensure target directory exists
os.makedirs(target_dir, exist_ok=True)
```

```python
pattern = r"Tilt_(\d+)_Azimuth"

# Traverse through subdirectories
for root, _, files in os.walk(source_dir):
    subfolder_name = os.path.basename(root)  # Get the subfolder name
    match = re.search(pattern,subfolder_name)
    if match:
        tx = match.group(1) #extract the tilt angle from the folder name

    # Traverse through files in the subdirectory
    for file in tqdm(files):
        if file.lower().endswith('tif'):  # Check if the file is an image
            old_path = os.path.join(root, file)
            # Create a new filename based on the subfolder and original name
            new_filename = f"TiltX_{tx}_TiltY_0_Thickness_{file}"
            new_path = os.path.join(target_dir, new_filename)
            
            # Copy the image to the new location
            shutil.copy2(old_path, new_path)
            #print(f"Copied: {old_path} -> {new_path}")
            
print("Renaming and moving completed.")
```
